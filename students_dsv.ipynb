# 4. Data Visualization

The following visualizations are used to explore the distributions, comparisons, relationships, and patterns present in the student performance dataset.

## Plot 1: Attendance Rate Distribution

This histogram shows the distribution of students' attendance rates and helps identify the most common attendance ranges.

```python
plt.figure(figsize=(8,5))

plt.hist(df["AttendanceRate"].dropna(), bins=10)

plt.title("Attendance Rate Distribution")
plt.xlabel("Attendance Rate")
plt.ylabel("Number of Students")

plt.show()
```

### Observation

The histogram shows how attendance rates are distributed among the students.

## Plot 2: Study Hours Per Week Distribution

This histogram shows how many hours students study per week.

```python
plt.figure(figsize=(8,5))

plt.hist(df["StudyHoursPerWeek"].dropna(), bins=10)

plt.title("Study Hours Per Week Distribution")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Number of Students")

plt.show()
```

### Observation

The plot shows the distribution of weekly study hours among students.

## Plot 3: Final Grade Distribution

This histogram shows the distribution of students' final grades.

```python
plt.figure(figsize=(8,5))

plt.hist(df["FinalGrade"].dropna(), bins=10)

plt.title("Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

### Observation

The visualization shows the different ranges of final grades obtained by students.

## Plot 4: Previous Grade Distribution

This histogram shows the distribution of students' previous grades.

```python
plt.figure(figsize=(8,5))

sns.histplot(df["PreviousGrade"].dropna(), kde=True)

plt.title("Previous Grade Distribution")
plt.xlabel("Previous Grade")
plt.ylabel("Frequency")

plt.show()
```

### Observation

The plot shows the distribution and overall pattern of previous grades.

## Plot 5: Gender Distribution

This bar chart shows the number of students in each gender category.

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Gender")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

## Plot 6: Parental Support Distribution

This pie chart shows the proportion of students receiving different levels of parental support.

```python
plt.figure(figsize=(7, 7))
df["ParentalSupport"].value_counts().plot(
    kind="pie",
        autopct="%1.1f%%"
)
plt.title("Parental Support Distribution")
plt.ylabel("")
plt.show()
```

## Plot 7: Online Classes Taken

This count plot shows the number of students according to whether they have taken online classes.

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="Online Classes Taken")

plt.title("Online Classes Taken")
plt.xlabel("Online Classes")
plt.ylabel("Number of Students")

plt.show()
```

## Plot 8: Extracurricular Activities

This bar chart shows the distribution of the extracurricular activity values recorded in the dataset.

```python
plt.figure(figsize=(7,5))

df["ExtracurricularActivities"].value_counts().sort_index().plot(kind="bar")

plt.title("Extracurricular Activities")
plt.xlabel("Extracurricular Activity Value")
plt.ylabel("Number of Students")

plt.show()
```

## Plot 9: Final Grade by Gender

This box plot compares the distribution of final grades between gender groups.

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x="Gender", y="FinalGrade")

plt.title("Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 10: Final Grade by Parental Support

This box plot compares final grades across different parental support levels.

```python
plt.figure(figsize=(9,5))

sns.boxplot(data=df, x="ParentalSupport", y="FinalGrade")

plt.title("Final Grade by Parental Support")
plt.xlabel("Parental Support")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 11: Attendance Rate by Gender

This violin plot compares the distribution of attendance rates across gender groups.

```python
plt.figure(figsize=(8,5))

sns.violinplot(data=df, x="Gender", y="AttendanceRate")

plt.title("Attendance Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Attendance Rate")

plt.show()
```

## Plot 12: Study Hours by Gender

This violin plot compares the distribution of weekly study hours across gender groups.

```python
plt.figure(figsize=(8,5))

sns.violinplot(data=df, x="Gender", y="StudyHoursPerWeek")

plt.title("Study Hours by Gender")
plt.xlabel("Gender")
plt.ylabel("Study Hours Per Week")

plt.show()
```

## Plot 13: Study Hours vs Final Grade

This scatter plot examines the relationship between weekly study hours and final grade.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
        x="StudyHoursPerWeek",
            y="FinalGrade"
            )

plt.title("Study Hours vs Final Grade")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 14: Attendance Rate vs Final Grade

This scatter plot examines whether attendance rate is related to final grade.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
        x="AttendanceRate",
            y="FinalGrade"
            )

plt.title("Attendance Rate vs Final Grade")
plt.xlabel("Attendance Rate")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 15: Previous Grade vs Final Grade

This scatter plot examines the relationship between previous grade and final grade.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
        x="PreviousGrade",
            y="FinalGrade"
            )

plt.title("Previous Grade vs Final Grade")
plt.xlabel("Previous Grade")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 16: Study Hours and Final Grade Regression

This regression plot shows the overall relationship between study hours and final grade.

```python
plt.figure(figsize=(8,5))

sns.regplot(
    data=df,
        x="StudyHoursPerWeek",
            y="FinalGrade",
                scatter_kws={"alpha": 0.5}
                )

plt.title("Study Hours vs Final Grade Regression")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 17: Attendance and Final Grade Regression

This regression plot shows the relationship between attendance rate and final grade.

```python
plt.figure(figsize=(8,5))

sns.regplot(
    data=df,
        x="AttendanceRate",
            y="FinalGrade",
                scatter_kws={"alpha": 0.5}
                )

plt.title("Attendance Rate vs Final Grade Regression")
plt.xlabel("Attendance Rate")
plt.ylabel("Final Grade")

plt.show()
```

## Plot 18: Average Final Grade by Gender

This bar chart compares the average final grade between gender groups.

```python
plt.figure(figsize=(7,5))

df.groupby("Gender")["FinalGrade"].mean().plot(kind="bar")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.xticks(rotation=0)

plt.show()
```

## Plot 19: Average Attendance by Parental Support

This chart compares average attendance rates across parental support levels.

```python
gender_performance = df.groupby("Gender")[
      ["PreviousGrade", "FinalGrade"]
].mean()

gender_performance.plot(kind="bar", figsize=(8,5))

plt.title("Average Academic Performance by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Grade")
plt.xticks(rotation=0)

plt.show()
```

## Plot 21: Online Classes by Gender

This stacked bar chart compares online class participation across gender groups.

```python
online_gender = pd.crosstab(
      df["Gender"],
          df["Online Classes Taken"]
          )

online_gender.plot(
              kind="bar",
                  stacked=True,
                      figsize=(8,5)
                      )

plt.title("Online Classes Taken by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.xticks(rotation=0)

plt.show()
```

## Plot 22: Average Final Grade by Study Hours

This line plot shows how average final grade changes across different study-hour values.

```python
study_grade = df.groupby("StudyHoursPerWeek")["FinalGrade"].mean()

plt.figure(figsize=(10,5))

plt.plot(study_grade.index, study_grade.values, marker="o")

plt.title("Average Final Grade by Study Hours")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Average Final Grade")

plt.show()
```

## Plot 23: Cumulative Final Grade Distribution

This area plot shows the cumulative distribution of final grades.

```python
grades = df["FinalGrade"].dropna().sort_values()

cumulative = np.arange(1, len(grades) + 1) / len(grades)

plt.figure(figsize=(9,5))

plt.fill_between(grades, cumulative, alpha=0.5)
plt.plot(grades, cumulative)

plt.title("Cumulative Final Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Cumulative Proportion")

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 24: Final Grade by Online Classes

This strip plot shows individual final-grade observations for different online class categories.

```python
plt.figure(figsize=(8,5))

sns.stripplot(
    data=df,
        x="Online Classes Taken",
            y="FinalGrade",
                jitter=True
                )

plt.title("Final Grade by Online Classes")
plt.xlabel("Online Classes Taken")
plt.ylabel("Final Grade")

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 25: Final Grade Distribution by Gender

This swarm plot displays individual final-grade observations for each gender group.

```python
plt.figure(figsize=(8,5))

sample_df = df.dropna(
    subset=["Gender", "FinalGrade"]
    ).sample(
        min(300, len(df.dropna(subset=["Gender", "FinalGrade"]))),
            random_state=42
            )

sns.swarmplot(
                data=sample_df,
                    x="Gender",
                        y="FinalGrade"
                        )

plt.title("Final Grade Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Final Grade")

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 26: Correlation Heatmap

This heatmap shows the correlation between the numerical variables in the dataset.

```python
plt.figure(figsize=(10,7))

numeric_df = df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
        annot=True,
            cmap="coolwarm",
                fmt=".2f"
                )

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 27: Pairwise Relationships

This pair plot provides an overall view of relationships among attendance, study hours, previous grade, and final grade.

```python
pair_data = df[
      [
              "AttendanceRate",
                      "StudyHoursPerWeek",
                              "PreviousGrade",
                                      "FinalGrade"
                                          ]
                                          ].dropna()

sns.pairplot(pair_data)

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 28: Missing Values by Column

This bar chart shows the number of missing values present in each column.

```python
plt.figure(figsize=(10,5))

missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0].sort_values(
    ascending=False
    )

missing_values.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 29: Average Final Grade by Online Classes

This bar chart compares the average final grade of students based on online classes taken.

```python
plt.figure(figsize=(8,5))

df.groupby("Online Classes Taken")["FinalGrade"].mean().plot(
    kind="bar"
    )

plt.title("Average Final Grade by Online Classes")
plt.xlabel("Online Classes Taken")
plt.ylabel("Average Final Grade")
plt.xticks(rotation=0)

plt.show()
```

### Observation

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.

## Plot 30: Overall Academic Performance

This histogram compares the distributions of previous grade and final grade to understand overall academic performance.

```python
plt.figure(figsize=(9,5))

plt.hist(
    df["PreviousGrade"].dropna(),
        bins=15,
            alpha=0.5,
                label="Previous Grade"
                )

plt.hist(
                    df["FinalGrade"].dropna(),
                        bins=15,
                            alpha=0.5,
                                label="Final Grade"
                                )

plt.title("Previous Grade vs Final Grade Distribution")
plt.xlabel("Grade")
plt.ylabel("Number of Students")
plt.legend()

plt.show()
```

The plot shows the distribution of the selected variable.
The most common values can be identified from the visualization.
