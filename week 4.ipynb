## Univariate Analysis

Univariate analysis examines one attribute at a time. It is used to understand the distribution and characteristics of individual variables.

Categorical attributes can be analyzed using frequency counts and bar charts, while numerical attributes can be analyzed using histograms and other distribution plots.

## Univariate Analysis

 ## Distribution of Job Titles

The `job_title_short` attribute is analyzed to understand the frequency of different job roles in the dataset.

```python
job_counts = df["job_title_short"].value_counts()

plt.figure(figsize=(10, 6))
job_counts.plot(kind="bar")
plt.title("Distribution of Job Titles")
plt.xlabel("Job Title")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

The bar chart shows the distribution of different job roles in the dataset. Some job roles have considerably more job postings than others, indicating differences in demand among the listed job categories.

## Distribution of Average Yearly Salary

The `salary_year_avg` attribute is analyzed to understand the distribution of yearly salaries among job postings.

A histogram is used because it shows how numerical salary values are distributed across different ranges.

```python
plt.figure(figsize=(10, 6))

plt.hist(df["salary_year_avg"].dropna(), bins=30)

plt.title("Distribution of Average Yearly Salary")
plt.xlabel("Average Yearly Salary")
plt.ylabel("Number of Job Postings")

plt.tight_layout()
plt.show()
```

### Observation

The histogram shows the distribution of average yearly salaries in the dataset. Most salary values are concentrated within a particular range, while relatively fewer job postings have very high salary values.

## Box Plot of Average Yearly Salary

A box plot is used to visualize the median, spread, and potential extreme values of the `salary_year_avg` attribute.

```python
plt.figure(figsize=(10, 5))

plt.boxplot(df["salary_year_avg"].dropna())

plt.title("Box Plot of Average Yearly Salary")
plt.ylabel("Average Yearly Salary")

plt.tight_layout()
plt.show()
```

## KDE Plot of Average Yearly Salary

A KDE (Kernel Density Estimate) plot is used to visualize the probability distribution of the `salary_year_avg` attribute.

It provides a smooth representation of how salary values are distributed.

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="salary_year_avg",
    fill=True
)

plt.title("KDE Plot of Average Yearly Salary")
plt.xlabel("Average Yearly Salary")
plt.ylabel("Density")

plt.tight_layout()
plt.show()
```

### Observation

The KDE plot shows the smooth distribution of average yearly salaries. The higher density indicates the salary ranges where more job postings are concentrated.

## Distribution of Job Schedule Types

The `job_schedule_type` attribute is analyzed to understand the different types of work schedules available in the job postings.

A bar chart is used to compare the frequency of each category.

```python
plt.figure(figsize=(12, 6))

schedule_counts = df["job_schedule_type"].value_counts().head(10)

schedule_counts.plot(kind="bar")

plt.title("Distribution of Job Schedule Types")
plt.xlabel("Job Schedule Type")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Observation

The bar chart shows that Full-time jobs form the largest category among the displayed job schedule types. Other categories such as Contractor, Internship, and Part-time occur less frequently.

## Distribution of Job Postings by Country

The `job_country` attribute is analyzed to understand the countries with the highest number of job postings.

A bar chart is used to compare the frequency of job postings across different countries.

```python
plt.figure(figsize=(12, 6))

country_counts = df["job_country"].value_counts().head(10)

country_counts.plot(kind="bar")

plt.title("Top 10 Countries by Number of Job Postings")
plt.xlabel("Country")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The bar chart shows the top 10 countries based on the number of job postings. The distribution indicates that job postings are concentrated in a few countries, while other countries have comparatively fewer postings.

## Work From Home Job Distribution

The `job_work_from_home` attribute is analyzed to compare job postings that allow work from home with those that do not.

A count plot is used to visualize the frequency of each category.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="job_work_from_home")

plt.title("Work From Home Job Distribution")
plt.xlabel("Work From Home")
plt.ylabel("Number of Job Postings")

plt.tight_layout()
plt.show()
```

### Observation

The count plot compares job postings based on whether work from home is available. It shows the difference in frequency between the two categories.

## Bivariate Analysis

Bivariate analysis examines the relationship between two attributes.

It helps identify patterns, differences, and relationships between variables in the dataset.

In this analysis, job roles and salary attributes are compared to understand salary differences across job categories.

## Average Yearly Salary by Job Title

The relationship between `job_title_short` and `salary_year_avg` is analyzed.

The average yearly salary is calculated for each major job role and visualized using a bar chart.

```python
salary_by_job = (
    df.groupby("job_title_short")["salary_year_avg"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12, 6))

salary_by_job.plot(kind="bar")

plt.title("Average Yearly Salary by Job Title")
plt.xlabel("Job Title")
plt.ylabel("Average Yearly Salary")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The bar chart compares the average yearly salary across different job roles. The average salary varies between job categories, showing that job role is associated with differences in the salary values reported in the dataset.

## Work From Home Availability by Job Title

The relationship between `job_title_short` and `job_work_from_home` is analyzed to compare work-from-home availability across different job roles.

A count plot is used to visualize the distribution.

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df,
    x="job_title_short",
    hue="job_work_from_home"
)

plt.title("Work From Home Availability by Job Title")
plt.xlabel("Job Title")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The count plot compares work-from-home availability across different job roles. The number of postings with and without work-from-home options varies between job categories.

## Job Postings by Country and Job Title

The relationship between `job_country` and `job_title_short` is analyzed to understand how different job roles are distributed across countries.

The top 10 countries and major job roles are selected to make the visualization easier to interpret.

```python
top_countries = df["job_country"].value_counts().head(10).index

country_job = df[
    df["job_country"].isin(top_countries)
]

plt.figure(figsize=(14, 7))

sns.countplot(
    data=country_job,
    x="job_country",
    hue="job_title_short"
)

plt.title("Job Postings by Country and Job Title")
plt.xlabel("Country")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The count plot shows the distribution of different job roles across the top 10 countries by number of job postings. The frequency of job roles varies between countries, indicating differences in the composition of job postings across locations.

## Distribution of Salary Rate

The `salary_rate` attribute is analyzed to understand the different payment periods reported in the job postings.

A bar chart is used to compare the frequency of each salary rate category.

```python
salary_rate_counts = df["salary_rate"].value_counts()

plt.figure(figsize=(8, 5))

salary_rate_counts.plot(kind="bar")

plt.title("Distribution of Salary Rate")
plt.xlabel("Salary Rate")
plt.ylabel("Number of Job Postings")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## Job Titles by Job Posting Source

The relationship between `job_title_short` and `job_via` is analyzed to understand how different job roles are distributed across major job posting sources.
